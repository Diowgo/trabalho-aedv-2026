```python
import pandas as pd
import altair as alt
```

###  Pergunta 1
#### Como a pontuação por jogo evolui ao longo do tempo

```python
df_Team_Stats_Per_Game = pd.read_csv("dados/Team Stats Per Game.csv")
```

```python
alt.Chart(df_Team_Stats_Per_Game).mark_point(size=25).encode(
    x=alt.X("season:Q", title="Temporada", scale=alt.Scale(zero=False)),
    y=alt.Y("pts_per_game:Q", title="Pontos por jogo", scale=alt.Scale(zero=False)),
    color=alt.Color(
        "playoffs:N",
        scale=alt.Scale(domain=[False, True], range=["orange", "crimson"]),
        title="Playoffs"
    ),
    shape="playoffs:N"
)

# Isso faz o altair tratar playoffs como nominal, permitindo que ele varie
```
